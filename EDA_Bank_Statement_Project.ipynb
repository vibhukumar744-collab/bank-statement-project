```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries successfully loaded!")
```

```python
df = pd.read_csv('cleaned_dataset.xls')
sns.set_theme(style="darkgrid")
df.head()
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='date', y='balance', color='teal', linewidth=2)
plt.title('Account Balance Over Time', fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['amount'], kde=True, color='crimson', bins=20)
plt.title('Transaction Amount Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Amount')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
top_desc = df['description'].value_counts().head(8)
sns.barplot(x=top_desc.values, y=top_desc.index, palette='magma')
plt.title('Top 8 Transaction Categories', fontsize=14, fontweight='bold')
plt.xlabel('Frequency')
plt.show()

print("\n--- KEY INSIGHTS ---")
print("1. Account balance trend shows key peaks and drop-offs over time.")
print("2. Most transaction amounts are clustered around small expenses.")
print("3. Top recurring descriptions highlight major spending channels.")
```
